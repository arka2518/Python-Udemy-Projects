```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
hlth = sns.load_dataset('healthexp')
hlth
```

```python
sns.lmplot(x = 'Year', y = 'Life_Expectancy', data = hlth)
```

```python
sns.lmplot(x = 'Year', y = 'Life_Expectancy', data = hlth, hue = 'Country', palette = 'coolwarm', markers = ['o', 'v', 'x', 's', 'D', '^'], 
          scatter_kws = {'s':50})
```
