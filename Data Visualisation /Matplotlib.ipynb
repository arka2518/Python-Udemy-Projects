```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Matplotlib Basics
```

```python
x = np.linspace(0,5,11)
x
y = x **3 
y
```

```python
plt.plot(x,y)
plt.title("Matplotlib")
plt.xlabel("Black")
plt.ylabel("Panther")
```

```python
# Subplots in Matplotlib
```

```python
# Graph 1
plt.subplot(2, 2, 1)
plt.plot(y,x)
plt.title("Graph 1")
# Graph 2
plt.subplot(2, 2, 2)
plt.plot(x,y)
plt.title("Graph 2")
# Graph 3
plt.subplot(2, 2, 3)
plt.plot(x,y)
plt.title("Graph 3")
# Graph 4
plt.subplot(2, 2, 4)
plt.plot(x**2,x**2)
plt.title("Graph 4")
```

```python
# Object Oriented way
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1, 0.1, 0.8, 1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2, 0.2, 0.6, 0.6])
axis2.plot(x,y)
```

```python
fig = plt.figure(figsize=(12,10), dpi=500)
axis1 = fig.add_axes([0.1, 0.1, 0.8, 1])
axis1.plot(x,y)
plt.title("Graph 1")
axis2 = fig.add_axes([0.2, 0.2, 0.6, 0.6])
axis2.plot(x,y)
plt.title("Graph 2")
```

```python
# Types of Plot
```

```python
plt.subplot(2, 2, 1)
axis1 = plt.scatter(x,y)
plt.subplot(2, 2, 2)
axis2 = plt.scatter(y,x)
plt.subplot(2, 2, 3)
axis3 = plt.scatter(x**2,x**2)
plt.subplot(2, 2, 4)
axis4 = plt.scatter(x,x)
```
