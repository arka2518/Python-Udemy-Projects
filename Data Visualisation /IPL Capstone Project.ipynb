```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings 
warnings.filterwarnings("ignore")

df = pd.read_csv( 'IPL.csv')
df.head()
```

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(x = match_wins.values, y = match_wins.index, palette = 'rainbow')
```

```python
df.head()
```

```python
# Toss Decision Trends
sns.countplot(x = df['toss_decision'], palette = 'viridis')
```

```python
# Toss Winner vs Match Winner
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]
percentage.round(2)
```

```python
# How do teams win? (Runs vs Wickets)
sns.countplot(x = df['won_by'])
```

```python
# Most "Player of the Match" Awards
player = df['player_of_the_match'].value_counts().head(10)
sns.barplot(x = player.values, y = player.index, palette = 'viridis')
```

```python
# Top Scorers
top_scorers = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending = False).head(10)
top_scorers.plot(kind = 'barh')
```

```python
# 10 Best Bowling Figures
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending = False).head(10)
sns.barplot(x = top_bowlers.values,y = top_bowlers.index, palette = 'mako')
```

```python
# Most Matches Played by Venue
most_matches = df['venue'].value_counts()
most_matches
sns.barplot(y = most_matches.index,x = most_matches.values,palette='rainbow')
```

```python
# Who won by the highest margin in runs
df[df['won_by'] == 'Runs'].sort_values(by = 'margin', ascending = False).head(1)[['match_winner', 'margin']]
```

```python
# Which player had the highest individual score
df[df['highscore'] == df['highscore'].max()][['top_scorer','highscore']]
```

```python
# Which bowler had the best bowling figures
df[df['highest_wickets'] == df['highest_wickets'].max()][['best_bowling', 'best_bowling_figure']]
```
