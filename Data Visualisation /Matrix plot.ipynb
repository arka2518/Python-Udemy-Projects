```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

## Heat map

```python
flights = sns.load_dataset('flights')
flights
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tip_corr = tips[['total_bill', 'tip', 'size']]
tip_corr
```

```python
tip_corr.corr()
```

```python
sns.heatmap(tip_corr.corr(), annot = True)
```

```python
sns.clustermap(tip_corr.corr(), annot = True)
```

## Pivot table Heat map

```python
pvt_flight = flights.pivot_table(values = 'passengers', index = 'month', columns = 'year', observed = 'False')
pvt_flight
```

```python
sns.heatmap(pvt_flight)
```

```python
sns.clustermap(pvt_flight)
```

```python
pvt_flight.corr()
```

```python
sns.heatmap(pvt_flight.corr())
```

```python
sns.clustermap(pvt_flight.corr())
```
