```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Categorical Data Plots
There are a few main plot types for this:

1. boxplot
2. violinplot
3. stripplot
4. swarmplot
5. barplot
6. countplot

```python
df = sns.load_dataset('titanic')
df
```

```python
sns.countplot(x = df['sex'], hue = df['alive'])
```

```python
sns.countplot(x = df['sex'], hue = df['age'])
```

```python
sns.barplot(x = df['sex'], y = df['survived'])
```

```python
sns.barplot(x = df['sex'], y = df['survived'], estimator = np.sum)
```

```python
sns.boxplot(x = 'age', y = 'deck', data = df)
```

```python
sns.violinplot(x = 'pclass', y = 'who', data = df)
```

```python
sns.stripplot(x = 'age', y = 'who', data = df)
```

```python
sns.swarmplot(x = 'age', y = 'who', data = df)
```

```python
sns.violinplot(x = 'age', y = 'who', data = df)
sns.swarmplot(x = 'age', y = 'who', data = df)
```
